```python
df["Month Number"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
```

```python
monthly_sales = (
    df.groupby(["Month Number", "Month Name"])["Sales"]
      .sum()
      .reset_index()
      .sort_values("Month Number")
)

monthly_sales
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales["Month Name"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    monthly_sales["Month Name"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=3
)

# Add value labels
for x, y in zip(monthly_sales["Month Name"], monthly_sales["Sales"]):
    plt.text(x, y+5000, f"{y:,.0f}", ha="center", fontsize=9)

plt.title("Monthly Sales Trend", fontsize=16, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales ($)", fontsize=12)

plt.xticks(rotation=45)

plt.grid(alpha=0.4)

plt.tight_layout()

plt.show()
```

```python
category_analysis = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Orders=("Order ID", "nunique")
      )
      .reset_index()
)

category_analysis
```

```python


plt.figure(figsize=(8,5))

plt.bar(
    category_analysis["Category"],
    category_analysis["Total_Sales"]
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    category_analysis["Category"],
    category_analysis["Total_Profit"]
)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

```python
top_customers = (
    df.groupby("Customer Name")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Orders=("Order ID", "nunique")
      )
      .sort_values(by="Total_Sales", ascending=False)
      .head(10)
      .reset_index()
)

top_customers
```

```python
plt.figure(figsize=(12,6))

plt.bar(top_customers["Customer Name"], top_customers["Total_Sales"])

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer Name")
plt.ylabel("Sales")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
region_analysis = (
    df.groupby("Region")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Orders=("Order ID", "nunique"),
          Customers=("Customer ID", "nunique")
      )
      .sort_values(by="Total_Sales", ascending=False)
      .reset_index()
)

region_analysis
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    region_analysis["Region"],
    region_analysis["Total_Sales"]
)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    region_analysis["Region"],
    region_analysis["Total_Profit"]
)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.show()
```

```python
state_analysis = (
    df.groupby("State")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
      .sort_values(by="Total_Sales", ascending=False)
      .reset_index()
)

state_analysis.head(10)
```

```python
plt.figure(figsize=(12,6))

plt.bar(state_analysis.head(10)["State"],
        state_analysis.head(10)["Total_Sales"])

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.bar(state_analysis.head(10)["State"],
        state_analysis.head(10)["Total_Profit"])

plt.title("Top 10 States by Profit")
plt.xlabel("State")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.axhline(0)

plt.show()
```

```python
discount_analysis = (
    df.groupby("Discount")
      .agg(
          Average_Profit=("Profit", "mean"),
          Total_Sales=("Sales", "sum"),
          Orders=("Order ID", "nunique")
      )
      .reset_index()
)

discount_analysis
```

```python
plt.figure(figsize=(10,6))

plt.plot(
    discount_analysis["Discount"],
    discount_analysis["Average_Profit"],
    marker="o"
)

plt.title("Average Profit by Discount")
plt.xlabel("Discount")
plt.ylabel("Average Profit")

plt.grid(True)
plt.show()
```

```python
segment_analysis = (
    df.groupby("Segment")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Customers=("Customer ID", "nunique"),
          Orders=("Order ID", "nunique")
      )
      .reset_index()
      .sort_values("Total_Sales", ascending=False)
)

segment_analysis
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    segment_analysis["Segment"],
    segment_analysis["Total_Sales"]
)

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")

plt.show()
```

```python
ship_analysis = (
    df.groupby("Ship Mode")
      .agg(
          Orders=("Order ID", "nunique"),
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
      .reset_index()
      .sort_values("Orders", ascending=False)
)

ship_analysis
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    ship_analysis["Ship Mode"],
    ship_analysis["Orders"]
)

plt.title("Orders by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Orders")

plt.xticks(rotation=15)

plt.show()
```

```python
subcategory_analysis = (
    df.groupby("Sub-Category")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
      .sort_values("Sales", ascending=False)
      .reset_index()
)

subcategory_analysis
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    subcategory_analysis["Sub-Category"],
    subcategory_analysis["Sales"]
)

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")

plt.xticks(rotation=60)

plt.tight_layout()

plt.show()
```

```python
top_products = (
    df.groupby("Product Name")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
      .sort_values("Sales", ascending=False)
      .head(10)
      .reset_index()
)

top_products
```

```python
plt.figure(figsize=(14,6))

plt.barh(
    top_products["Product Name"],
    top_products["Sales"]
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")

plt.tight_layout()

plt.show()
```
